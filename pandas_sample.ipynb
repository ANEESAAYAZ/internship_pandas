```python
orders_with_total=orders.assign(revenue=orders["quantity"] * orders["unit_price"])
```

```python
orders.head()
```

```python
orders_with_total
```

Sorting

```python
orders_with_total.sort_values(by="revenue",ascending=False).head()
```

Group By

```python
products=pd.read_csv("products.csv")
```

```python
products.head()
```

```python
enriched=orders_with_total.merge(products,on="product_id")
```

```python
enriched.head()
```

```python
revenue_by_category=enriched.groupby("category")["revenue"].sum()
```

```python
revenue_by_category
```

```python
category_count=enriched.groupby("category")["revenue"].count()
```

```python
category_count
```

```python
multi_agg=enriched.groupby("category").agg(
    revenue_sum=("revenue","sum"),
    revenue_mean=("revenue","mean"),
    order_count=("revenue","count")
)
```

```python
multi_agg
```

```python
customers=pd.read_csv("customers (1).csv")
```

```python
customers.head()
```

```python
full=orders_with_total.merge(customers,on="customer_id",how="left").merge(products,on="product_id",how="left")
```

```python
full.head()
```

Date operations

```python
full["order_month"]=full["order_date"].dt.month
```

```python
full["order_quarter"]=full["order_date"].dt.quarter
```

```python
full.head()
```

```python
full["order_quarter"].value_counts()
```

```python
monthly_revenue=full.groupby(["order_month","order_quarter"])["revenue"].sum()
```

```python
monthly_revenue
```

```python
full.groupby(["order_month"])["revenue"].sum().sort_values(ascending=False)
```

```python
country_category=full.pivot_table(index="country",columns="category",values="revenue",aggfunc="sum")
```

```python
country_category
```

```python
full.to_csv("full.csv")
```
